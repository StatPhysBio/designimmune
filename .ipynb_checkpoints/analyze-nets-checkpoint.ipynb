# Analyze and plot results for networks

This script takes in the results from run_over_regs.sh, assembles them into a single dataset, and performs analyses on the data.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

import matplotlib as mpl
import matplotlib.ticker as ticker

import numpy as np
import pandas as pd
from pydmd import DMD
import os
import plotly.graph_objects as go
import plotly.express as px

# import clustering packages
from sklearn.preprocessing import StandardScaler, RobustScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
import seaborn as sns
from celluloid import Camera

plt.style.use('custom.mplstyle')
%config InlineBackend.figure_format = 'retina'
from tqdm import tqdm

from stoch_sim_model import *
```

## 0. Load data and build datasets

```python
# Load data from 2nd infections
sim_kind = 'agent'
infection_type = 'sec'
reg_model = ''
runs = '-10-'
comment = "act-reg-exp-reg" #'mem-reg'

d_mean = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/summary_stats/mean/stacked_data'+runs+'runs'+'-'+comment+'.npy'
d_std = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/summary_stats/std/stacked_data'+runs+'runs'+'-'+comment+'.npy'


mean_data = np.load(d_mean)
std_data = np.load(d_std)

# create dataframe of infection response statistics
var_names = np.concatenate((param_names_for_df, stat_names_for_df))
mean_df = pd.DataFrame(mean_data, columns = [i for i in var_names]).dropna(thresh=1)
std_df = pd.DataFrame(std_data, columns = [i for i in var_names])

with pd.option_context('display.max_columns', None):
    display(mean_df)
```

## 1. Understanding the statistics of responses to an infection

```python
# Create additional variables
K_IEs = np.unique(mean_df['K_IE'])
d_Is = np.unique(mean_df['d_I'])
b_S = np.mean(mean_df['d_S']*mean_df['S_0'])
N_0 = np.mean(mean_df['N_0'])

mean_df['harm_pI_scaled'] = mean_df['harm_pI']/(b_S)
mean_df['harm_sI_scaled'] = mean_df['harm_sI']/(b_S)
mean_df['harm_pS_scaled'] = mean_df['harm_pS']/(b_S)
mean_df['harm_sS_scaled'] = mean_df['harm_sS']/(b_S)
mean_df['harm_pIS'] = (mean_df['harm_pI'] + mean_df['harm_pS'])
mean_df['harm_sIS'] = (mean_df['harm_sI'] + mean_df['harm_sS'])
mean_df['harm_pIS_scaled'] = (mean_df['harm_pI'] + mean_df['harm_pS'])/(b_S)
mean_df['harm_sIS_scaled'] = (mean_df['harm_sI'] + mean_df['harm_sS'])/(b_S)
mean_df['harm_totIS_scaled'] = (mean_df['harm_pI'] + mean_df['harm_pS'] + mean_df['harm_sI'] + mean_df['harm_sS'])/(b_S)
mean_df['max_pE_scaled'] = mean_df['max_pE']/mean_df['K_IE']
mean_df['max_sE_scaled'] = mean_df['max_sE']/mean_df['K_IE']
mean_df['int_pE_scaled'] = mean_df['int_pE']/mean_df['K_IE']
mean_df['int_sE_scaled'] = mean_df['int_sE']/mean_df['K_IE']
mean_df['inf_pmem'] = (mean_df['inf_pM'] + mean_df['inf_pM'])
mean_df['inf_smem'] = (mean_df['inf_sM'] + mean_df['inf_sM'])
mean_df['inf_pmem_scaled'] = (mean_df['inf_pM'] + mean_df['inf_pM'])/mean_df['K_IE']
mean_df['inf_smem_scaled'] = (mean_df['inf_sM'] + mean_df['inf_sM'])/mean_df['K_IE']
mean_df['inf_pM_scaled'] = mean_df['inf_pM']/mean_df['K_IE']
mean_df['inf_pM_scaled'] = mean_df['inf_sM']/mean_df['K_IE']
mean_df['inf_sM_scaled'] = mean_df['inf_pM']/mean_df['K_IE']
mean_df['init_M_scaled'] = mean_df['init_M']/mean_df['K_IE']
mean_df['inf_sM_scaled'] = mean_df['inf_sM']/mean_df['K_IE']
mean_df['harm_I_fld_chg'] = mean_df['min_sS']/mean_df['min_pS']
mean_df['harm_I_diff_chg'] = (mean_df['min_sS']-mean_df['min_pS'])/mean_df['min_pS']
mean_df['max_presp_scaled'] = mean_df['max_pE_scaled'] + mean_df['inf_pM_scaled']
mean_df['max_sresp_scaled'] = mean_df['max_sE_scaled'] + mean_df['inf_sM_scaled']
mean_df['max_totresp_scaled'] = mean_df['max_presp_scaled'] + mean_df['max_sresp_scaled']
mean_df['int_presp_scaled'] = mean_df['int_pE_scaled'] + mean_df['inf_pM_scaled']
mean_df['int_sresp_scaled'] = mean_df['int_sE_scaled'] + mean_df['inf_sM_scaled']
mean_df['int_totresp_scaled'] = mean_df['int_presp_scaled'] + mean_df['max_sresp_scaled']
mean_df['max_presp'] = mean_df['max_pE'] + mean_df['inf_pM']
mean_df['max_sresp'] = mean_df['max_sE'] + mean_df['inf_sM']
mean_df['max_totresp'] = mean_df['max_presp'] + mean_df['max_sresp']
mean_df['int_presp'] = mean_df['int_pE'] + mean_df['inf_pM']
mean_df['int_sresp'] = mean_df['int_sE'] + mean_df['inf_sM']
mean_df['int_totresp'] = mean_df['int_presp'] + mean_df['max_sresp']
mean_df['d_I_K_IE'] = mean_df.groupby(['d_I', 'K_IE'], as_index=False).ngroup()

mean_df['no_reg'] = (mean_df['psi_NM_I'] + mean_df['psi_NM_H'] + mean_df['psi_NM_IH'])*1 == 0

# identify Biologically evidenced networks
mean_df['bio_reg'] = (mean_df['psi_NM_I'] > 0.0)*(mean_df['psi_NM_H'] > 0.0)*(mean_df['psi_NM_IH'] > 0.0)*1

std_df['harm_pI_scaled'] = std_df['harm_pI']/(b_S)
std_df['harm_sI_scaled'] = std_df['harm_sI']/(b_S)
std_df['harm_pS_scaled'] = std_df['harm_pS']/(b_S)
std_df['harm_sS_scaled'] = std_df['harm_sS']/(b_S)
std_df['harm_pIS_scaled'] = np.sqrt(std_df['harm_pI']**2 + std_df['harm_pS']**2)/(b_S)
std_df['harm_sIS_scaled'] = np.sqrt(std_df['harm_sI']**2 + std_df['harm_sS']**2)/(b_S)
std_df['max_pE_scaled'] = std_df['max_pE']/mean_df['K_IE']
std_df['max_sE_scaled'] = std_df['max_sE']/mean_df['K_IE']
std_df['int_pE_scaled'] = std_df['int_pE']/mean_df['K_IE']
std_df['int_sE_scaled'] = std_df['int_sE']/mean_df['K_IE']

vir_prop = np.unique(mean_df[['d_I','K_IE']].to_numpy(), axis = 0)
NM_reg = ['psi_NM_I', 'psi_NM_H', 'psi_NM_IH']
EM_reg = ['psi_EM_I', 'psi_EM_H', 'psi_EM_IH']
Nact_reg = ['psi_Nact_I', 'psi_Nact_H', 'psi_Nact_IH']
Ediv_reg = ['psi_Ediv_I', 'psi_Ediv_H', 'psi_Ediv_IH']
reg = Ediv_reg #Nact_reg #+ Ediv_reg
```

#### (a) Effects of infection parameters on immune response are non-linear
By plotting the relationship between 1st and 2nd infection harm for specific infections, $(d_I, K_{I,E})$ pairs, and computing the magnitude of cell death incurred, as $\frac{S_p^{min}}{S_0}$, we can see that there is a range of possible responses depending on the range of infections.

```python
# Plot trends for different networks based on infection severity
virs = vir_prop
n = len(virs)
rows = 1
fig, axs = plt.subplots(rows, int(np.ceil(n/rows)), figsize=(4*rows*n, n), sharex=False, sharey='row')

for l, (d_I, K_IE) in enumerate(virs):
    
    x,y = int(np.floor(rows*l/n)), l % int(np.ceil((n/rows)))
    
    data_mean = mean_df[(mean_df['d_I'] == d_I)*(mean_df['K_IE'] == K_IE)]
    data_std = std_df[(mean_df['d_I'] == d_I)*(mean_df['K_IE'] == K_IE)]
    
    S_min_over_S_0 = np.round(np.max(data_mean['min_pS']/data_mean['S_0']),2)
    var = 'harm_sIS_scaled' # change this to look at different variables
    im = axs[y].scatter(data_mean[var], data_std[var]/data_mean[var], 
                          c = data_std['max_pE_scaled'], cmap='coolwarm_r',
                          s = 15, marker ='s', alpha = 0.2)
    
    axs[y].set_title(r"$d_I =$ "+str(int(d_I/d_S))+r"$\times d_S$"+r", $K_{IE} =$ "+str(np.round(K_IE/K_SE,5))+r"$\times K_{S,E}$"+"\n"+r"$min_{\psi}~\{ S_{min}/S_0 \} =$ "+str(S_min_over_S_0), fontsize =7.0)

fig.subplots_adjust(wspace=.25, hspace=.0)
cb = fig.colorbar(im, ax=axs, orientation='vertical', shrink = 0.6)
cb.set_label("1st effector response", fontsize = 12)

for j, ax in enumerate(axs.flat):
    if j % int(np.ceil(n/rows)) == 0:
        ax.set(ylabel=r"$\frac{Standard~deviation}{Mean}$")
    # if j >= int(np.ceil(n/rows)):
    ax.set(xlabel=r"$Mean$")
    
plt.savefig('_figs/noise_in_'+var+'.png', dpi=300, bbox_inches='tight')
```

## 4. Performance-based clustering of immune networks

```python
# normalize data
data_scaler = MinMaxScaler()
num_clusters = 6
cutoff = 0.95
infection_scenarios = []
no_eff_data = [[] for i in np.arange(len(vir_prop))]

for l, (d_I, K_IE) in enumerate(vir_prop):
    data = mean_df.loc[(mean_df["d_I"] == d_I)*(mean_df["K_IE"] == K_IE)*(mean_df["K_EI"] == K_IE if d_I > 0 else True)*(mean_df["K_EH"] == mean_df["K_IH"]),]
    std_data = std_df.loc[(mean_df["d_I"] == d_I)*(mean_df["K_IE"] == K_IE)*(mean_df["K_EI"] == K_IE if d_I > 0 else True)*(mean_df["K_EH"] == mean_df["K_IH"]),]

    # compute infection harm without T cell response
    no_eff_data[l] = lin_stoch_sim(N_0 = 0, K_IE = K_IE, d_I = d_I)
    no_eff_stats = no_eff_data[l]["summary_stats"]
    
    data.loc[:,"peff_protection"] = (no_eff_stats[4] - data['harm_pI'].to_numpy())/(b_S)
    data.loc[:,"seff_protection"] = (no_eff_stats[5] - data['harm_sI'].to_numpy())/(b_S)

    data.loc[:,"peff_toxicity"] = (data['harm_pS'].to_numpy() - 0*no_eff_stats[6])/(b_S)
    data.loc[:,"seff_toxicity"] = (data['harm_sS'].to_numpy() - 0*no_eff_stats[7])/(b_S)

    data.loc[:,"peff_utility"] = (no_eff_stats[4] + 1.0*no_eff_stats[6] - data['harm_pIS'].to_numpy())/(b_S)
    data.loc[:,"seff_utility"] = (no_eff_stats[5] + 1.0*no_eff_stats[7] - data['harm_sIS'].to_numpy())/(b_S)

    data.loc[:,"peff_efficiency"] = (data['peff_protection'] > 0)*(data['peff_protection'] - data['peff_toxicity'])/data['peff_protection']
    data.loc[:,"seff_efficiency"] = (data['seff_protection'] > 0)*(data['seff_protection'] - data['seff_toxicity'])/data['seff_protection']
    
    data.loc[:,"peff_protection_frac"] = 1 - data['harm_pI'].to_numpy()/(no_eff_stats[-20])
    data.loc[:,"seff_protection_frac"] = 1 - data['harm_sI'].to_numpy()/(no_eff_stats[-19])
    
    data.loc[:,"peff_protection_log"] = np.log((no_eff_stats[4])/data['harm_pI'].to_numpy())
    data.loc[:,"seff_protection_log"] = np.log((no_eff_stats[6])/data['harm_sI'].to_numpy())

    data.loc[:,"K_EI_scaled"] = data['K_EI']/K_IE
    data.loc[:,"K_EH_scaled"] = data['K_EH']/K_IH

    data.loc[:, "high_putility"] = 1*(data['peff_utility'] >= np.quantile(data['peff_utility'], cutoff))
    data.loc[:, "high_pprotection"] = 1*(data['peff_protection'] >= np.quantile(data['peff_protection'], cutoff))
    data.loc[:, "high_presponse"] = 1*(data['int_presp'] >= np.quantile(data['int_presp'], cutoff))
    data.loc[:, "high_pefficiency"] = 1*(data['peff_efficiency'] >= np.quantile(data['peff_efficiency'], cutoff))
    data.loc[:, "high_sutility"] = 1*(data['seff_utility'] >= np.quantile(data['seff_utility'], cutoff))
    data.loc[:, "high_sprotection"] = 1*(data['seff_protection'] >= np.quantile(data['seff_protection'], cutoff))
    data.loc[:, "high_sresponse"] = 1*(data['int_sresp'] >= np.quantile(data['int_sresp'], cutoff))
    data.loc[:, "high_sefficiency"] = 1*(data['seff_efficiency'] >= np.quantile(data['seff_efficiency'], cutoff))
    
    infection_scenarios.append(data)

# stack datasets
clustered_mean_df = pd.concat(infection_scenarios)

# helper variables
perf_categories = ["bio_reg", "high_putility", "high_pprotection", "high_pefficiency", "high_presponse", 
                   "high_sutility", "high_sprotection", "high_sefficiency", "high_sresponse"]
perf_vars = ["peff_utility", "peff_protection", "peff_efficiency", "int_presp", 
             "seff_utility", "seff_protection", "seff_efficiency", "int_sresp"]
perf_cat_labels = ['BM','pHU','pHE1','pHE2','pHR', 'sHU','sHE1', 'sHE2', 'sHR']
perf_labels = ["1st response \n utility", "1st response \n protection", "1st response \n efficiency", "1st response \n magnitude",
               "2nd response \n utility", "2nd response \n protection", "2nd response \n efficiency", "2nd response \n magnitude"]


num_categories = len(perf_categories)
chosen_virs = np.unique(vir_prop[:,[0,1]], axis = 0)
n = len(chosen_virs)
```

## 5. Risk-protection Trade-off

```python
# plot memory and effector response against infection excess death
fig, axs = plt.subplots(2, 2, figsize=(10, 9)) #, sharex=True, sharey=True)
#reg = ['psi_NM_I', 'psi_NM_H', 'psi_NM_IH', 'psi_EeM_I', 'psi_EeM_H', 'psi_EeM_IH', 'psi_pME_I', 'psi_pME_H', 'psi_pME_IH']

p_reward_data = clustered_mean_df.groupby(p_reg).mean()
p_risk_data = clustered_mean_df.groupby(p_reg).std()/p_reward_data

s_reward_data = clustered_mean_df.groupby(s_reg).mean()
s_risk_data = clustered_mean_df.groupby(s_reg).std()/s_reward_data
var = 'eff_protection'

axs[0,0].scatter(p_reward_data['p'+var], p_risk_data['p'+var], marker = '.', color = 'black', alpha = 0.05)
axs[0,0].scatter(p_reward_data['p'+var][p_reward_data['bio_reg'] > 0.0],
                p_risk_data['p'+var][p_reward_data['bio_reg'] > 0.0], 
                color ='gold', s = 30, marker ='X', zorder=2, edgecolor = 'k', linewidth = 0.3, label = "Experimentally motivated")
#axs[0,0].semilogy()

axs[0,1].scatter(p_reward_data["int_presp"], p_risk_data["int_presp"], marker = '.', color = 'black', alpha = 0.05)
axs[0,1].scatter(p_reward_data["int_presp"][p_reward_data['bio_reg'] > 0.0],
                p_risk_data["int_presp"][p_reward_data['bio_reg'] > 0.0], 
                color ='gold', s = 30, marker ='X', zorder=2, edgecolor = 'k', linewidth = 0.3)
axs[0,1].semilogx()

axs[1,0].scatter(s_reward_data['s'+var], s_risk_data['s'+var], marker = '.', color = 'black', alpha = 0.05)
axs[1,0].scatter(s_reward_data['s'+var][s_reward_data['bio_reg'] == 1],
                s_risk_data['s'+var][s_reward_data['bio_reg'] == 1], 
                color ='gold', s = 30, marker ='X', zorder=2, edgecolor = 'k', linewidth = 0.3)
#axs[1,0].semilogy()

axs[1,1].scatter(s_reward_data["int_sresp"], s_risk_data["int_sresp"], marker = '.', color = 'black', alpha = 0.05)
axs[1,1].scatter(s_reward_data["int_sresp"][s_reward_data['bio_reg'] == 1],
                s_risk_data["int_sresp"][s_reward_data['bio_reg'] == 1], 
                color ='gold', s = 30, marker ='X', zorder=2, edgecolor = 'k', linewidth = 0.3)
axs[1,1].semilogx()

axs[0,0].legend()

titles = ["1st infection protection \n (units of healthy death)", "1st infection responsiveness \n (cells)",
         "2nd infection protection \n (units of healthy death)", "2nd infection responsiveness \n (cells)"]
for i, ax in enumerate(axs.flat):
    if i % 2 == 0:
        ax.set(ylabel=r"$\frac{Standard~deviation}{Mean}$")
    if i >= 2:
        ax.set(xlabel=r"$Mean$")
    ax.set_title(titles[i], fontsize = 9)
    #ax.legend()
        
plt.savefig('_figs/risk_reward_tradeoff'+'.png', dpi=300, bbox_inches='tight')
```

## 6. Investigating cell fate decisions within groups

```python
d_p_diff_bias = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/prim_diff_bias/mean/'
d_s_diff_bias = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/sec_diff_bias/mean/'
d_sim_sum = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/summary_stats/mean/'
d_dyn = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/cell_time_series/mean/'

infection_type = 'sec'
file_list = [f for f in os.listdir(d_p_diff_bias) if runs in f and infection_type in f and comment in f]
reg = ['psi_NM_I', 'psi_NM_H', 'psi_NM_IH','psi_EeM_I', 'psi_EeM_H','psi_EeM_IH','psi_pME_I', 'psi_pME_H','psi_pME_IH']

pnts = 500
n = len(vir_prop)
fig, axs = plt.subplots(n,num_categories, figsize=(2*n, 2*num_categories), sharex=True, sharey=True)
p_diff_data_for_dmd = [[] for i in np.arange(0, n*num_categories)]
s_diff_data_for_dmd = [[] for i in np.arange(0, n*num_categories)]
dyn_data_for_dmd = [[] for i in np.arange(0, n*num_categories)]
p_diff_data_for_dmd_sub = [[] for i in np.arange(0, n*num_categories)]
s_diff_data_for_dmd_sub = [[] for i in np.arange(0, n*num_categories)]
dyn_data_for_dmd_sub = [[] for i in np.arange(0, n*num_categories)]
data = [clustered_mean_df[(clustered_mean_df['d_I'] == d_I)*(clustered_mean_df['K_IE'] == K_IE)] for (d_I, K_IE) in vir_prop]
keep = [i*int(sim_steps/pnts) for i in np.arange(0,pnts)]
time = np.linspace(0, sim_duration, int(sim_steps) + 1)[keep]

for file_count, f in enumerate(tqdm(file_list)):
    p_diff_bias_net = np.load(os.path.join(d_p_diff_bias, f))
    s_diff_bias_net = np.load(os.path.join(d_s_diff_bias, f))
    dyn_net = np.load(os.path.join(d_dyn, f))
    sim_sum_net = np.load(os.path.join(d_sim_sum, f))[0][-33:-24] # network parameters

    for l, (d_I, K_IE) in enumerate(vir_prop):
        p_diff_bias = p_diff_bias_net[l][keep]
        s_diff_bias = s_diff_bias_net[l][keep]
        dyn = dyn_net[l][keep]

        for i, category in enumerate(perf_categories):
            if  list(sim_sum_net) in [list(i) for i in data[l].loc[data[l][category] == 1, reg].to_numpy()]:
                axs[l,i].plot(time, p_diff_bias[:,0], color='black', 
                            label = r'Bias$_t(N^* \longrightarrow E)$', alpha = 0.1)
                axs[l,i].plot(time, p_diff_bias[:,2], color='red',
                            label = r'Bias$_t(E \longrightarrow eM)$', alpha = 0.1)
                # save raw data
                diff_index = i + num_categories*l
                p_diff_data_for_dmd[diff_index].append(p_diff_bias)
                s_diff_data_for_dmd[diff_index].append(s_diff_bias)
                dyn_data_for_dmd[diff_index].append(dyn)

            if  list(sim_sum_net) in [list(i) for i in data[l].loc[data[l][category] == 0, reg].to_numpy()]:
                diff_index = i + num_categories*l
                p_diff_data_for_dmd_sub[diff_index].append(p_diff_bias)
                s_diff_data_for_dmd_sub[diff_index].append(s_diff_bias)
                dyn_data_for_dmd_sub[diff_index].append(dyn)
            
        axs[l,0].set_title(r"$d_I =$ "+str(np.round(d_I,3))+r", $K_{I,E} =$ "+str(np.round(K_IE,0)), fontsize = 8)
        
        # if file_count == 0:
        #     axs[0,0].legend(fontsize = 8)
        #     axs[1,0].legend(fontsize = 8)
        #     axs[2,0].legend(fontsize = 8)
    
for i, ax in enumerate(axs.flat):
    ax.set(xlabel= r"time, $t$ (days)")

for ax in axs.flat:
    ax.label_outer()
        
plt.savefig('_figs/pop_diff_dynamics_by_category', dpi=300, bbox_inches='tight')
```
