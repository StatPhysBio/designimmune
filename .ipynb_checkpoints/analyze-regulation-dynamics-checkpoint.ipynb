This script takes in the results from run_over_regs.sh, assembles them into a single dataset, and performs analyses on the data.

## 0. Load data and build datasets

```python
# Load data from 2nd infections
sim_kind = 'agent'
infection_type = 'sec'
reg_model = ''
runs = '-10-'
comment = "act-reg-exp-reg" #'mem-reg'

d_mean = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/summary_stats/mean/processed'+runs+'runs'+'-'+comment+'.npy'
d_std = '/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/summary_stats/std/processed'+runs+'runs'+'-'+comment+'.npy'


mean_df = np.load(d_mean)
std_df = np.load(d_std)

with pd.option_context('display.max_columns', None):
    display(mean_df)
```

## 1. Performance-based clustering of immune networks

```python
# normalize data
data_scaler = MinMaxScaler()
num_clusters = 6
cutoff = 0.995
infection_scenarios = []
no_eff_data = [[] for i in np.arange(len(virs))]

for l, (d_I, K_IE) in enumerate(virs):
    data = mean_df.loc[(mean_df["d_I"] == d_I)*(mean_df["K_IE"] == K_IE)*(mean_df["K_EI"] == K_IE if d_I > 0 else True)*(mean_df["K_EH"] == mean_df["K_IH"]),]
    std_data = std_df.loc[(mean_df["d_I"] == d_I)*(mean_df["K_IE"] == K_IE)*(mean_df["K_EI"] == K_IE if d_I > 0 else True)*(mean_df["K_EH"] == mean_df["K_IH"]),]

    # compute infection harm without T cell response
    no_eff_data[l] = lin_stoch_sim(N_0 = 0, K_IE = K_IE, d_I = d_I)
    no_eff_stats = no_eff_data[l]["summary_stats"]
    
    data.loc[:,"peff_protection"] = (no_eff_stats[4] - data['harm_pI'].to_numpy())/(b_S)
    data.loc[:,"seff_protection"] = (no_eff_stats[5] - data['harm_sI'].to_numpy())/(b_S)

    data.loc[:,"peff_toxicity"] = (data['harm_pS'].to_numpy() - 0*no_eff_stats[6])/(b_S)
    data.loc[:,"seff_toxicity"] = (data['harm_sS'].to_numpy() - 0*no_eff_stats[7])/(b_S)

    data.loc[:,"peff_utility"] = (no_eff_stats[4] + 1.0*no_eff_stats[6] - data['harm_pIS'].to_numpy())/(b_S)
    data.loc[:,"seff_utility"] = (no_eff_stats[5] + 1.0*no_eff_stats[7] - data['harm_sIS'].to_numpy())/(b_S)

    data.loc[:,"peff_efficiency"] = (data['peff_protection'] > 0)*(data['peff_protection'] - data['peff_toxicity'])/data['peff_protection']
    data.loc[:,"seff_efficiency"] = (data['seff_protection'] > 0)*(data['seff_protection'] - data['seff_toxicity'])/data['seff_protection']
    
    data.loc[:,"peff_protection_frac"] = 1 - data['harm_pI'].to_numpy()/(no_eff_stats[-20])
    data.loc[:,"seff_protection_frac"] = 1 - data['harm_sI'].to_numpy()/(no_eff_stats[-19])
    
    data.loc[:,"peff_protection_log"] = np.log((no_eff_stats[4])/data['harm_pI'].to_numpy())
    data.loc[:,"seff_protection_log"] = np.log((no_eff_stats[6])/data['harm_sI'].to_numpy())

    data.loc[:, "high_putility"] = 1*(data['peff_utility'] >= np.quantile(data['peff_utility'], cutoff))
    data.loc[:, "high_ptoxicity"] = 1*(data['peff_protection'] >= np.quantile(data['peff_toxicity'], cutoff))
    data.loc[:, "high_presponse"] = 1*(data['int_presp'] >= np.quantile(data['int_presp'], cutoff))
    data.loc[:, "high_pefficiency"] = 1*(data['peff_efficiency'] >= np.quantile(data['peff_efficiency'], cutoff))
    data.loc[:, "high_sutility"] = 1*(data['seff_utility'] >= np.quantile(data['seff_utility'], cutoff))
    data.loc[:, "high_stoxicity"] = 1*(data['seff_protection'] >= np.quantile(data['seff_toxicity'], cutoff))
    data.loc[:, "high_sresponse"] = 1*(data['int_sresp'] >= np.quantile(data['int_sresp'], cutoff))
    data.loc[:, "high_sefficiency"] = 1*(data['seff_efficiency'] >= np.quantile(data['seff_efficiency'], cutoff))
    
    infection_scenarios.append(data)

# stack datasets
clustered_mean_df = pd.concat(infection_scenarios)

# helper variables
perf_categories = ["high_putility", "high_ptoxicity", "high_pefficiency", "high_presponse", 
                   "high_sutility", "high_stoxicity", "high_sefficiency", "high_sresponse"]
perf_vars = ["peff_utility", "peff_toxicity", "peff_efficiency", "int_presp", 
             "seff_utility", "seff_toxicity", "seff_efficiency", "int_sresp"]
perf_labels = ["1st response \n utility", "1st response \n toxicity", "1st response \n efficiency", "1st response \n magnitude",
               "2nd response \n utility", "2nd response \n toxicity", "2nd response \n efficiency", "2nd response \n magnitude"]


num_categories = len(perf_categories)
chosen_virs = virs[1:]
n = len(chosen_virs)
```

```python
# Plot trends for different networks based on infection severity
n_virs = len(chosen_virs)
reg = Nact_reg + Ediv_reg
fig, axs = plt.subplots(n_virs, 4, figsize=(n_virs**2, n_virs**2), sharex=True, sharey=True)
z_varname = 'peff_utility'

for l, (d_I, K_IE) in enumerate(chosen_virs):
    vir_data = clustered_mean_df.loc[(clustered_mean_df['d_I'] == d_I) & (clustered_mean_df['K_IE'] == K_IE)]
    
    x = l % n_virs
    for y in np.arange(0,4):
        if y == 0:
            data = vir_data[clustered_mean_df[reg[2]] >= 0.0].groupby(reg[0:3], as_index=False).mean()
            x_varname, y_varname = reg[0:2]
            z_label = r"$\psi_{N^*}^{(I,H)}$ "+("$\geq 0$,")
        elif y == 1:
            data = vir_data[clustered_mean_df[reg[2]] <= 0.0].groupby(reg[0:3], as_index=False).mean()
            x_varname, y_varname = reg[0:2]
            z_label = r"$\psi_{N^*}^{(I,H)}$ "+("$\leq 0$,")
        elif y == 2:
            data = vir_data[clustered_mean_df[reg[-1]] >= 0.0].groupby(reg[-3:], as_index=False).mean()
            x_varname, y_varname = reg[-3:-1]
            z_label = r"$\psi_{E*}^{(I,H)}$ "+("$\geq 0$,")
        elif y == 3:
            data = vir_data[clustered_mean_df[reg[-1]] <= 0.0].groupby(reg[-3:], as_index=False).mean()
            x_varname, y_varname = reg[-3:-1]
            z_label = r"$\psi_{E*}^{(I,H)}$ "+("$\leq 0$,")
            
        S_min_over_S_0 = np.round(np.max(data['min_pS']/data['S_0']),2)
        vmax= max(vir_data[z_varname])
        vmin= min(vir_data[z_varname])
        
        axs[x,y].tricontour(data[x_varname], data[y_varname], data[z_varname], linewidths=0.2, colors='k', #levels = 3,
                                  vmax= vmax, vmin= vmin)
        im = axs[x,y].tricontourf(data[x_varname], data[y_varname], data[z_varname], cmap="coolwarm_r", #levels = 3,
                                  vmax= vmax, vmin= vmin)
        axs[x,y].set_title(z_label+"\n"+"$d_I =$ "+str(int(d_I/d_S))+r"$\times d_S$"+r", $K_{IE} =$ "+str(np.round(K_IE/K_SE,5))+r"$\times K_{S,E}$"+"\n"+r"$min_{\psi}~\{ S_{min}/S_0 \} =$ "+str(S_min_over_S_0), fontsize =10.0)
        axs[x,y].set_aspect('equal', adjustable='box')
        cb = fig.colorbar(im, ax=axs[x,y], orientation='vertical', shrink = 0.75, pad = 0.075)
        cb.set_label("1st response utility", fontsize = 10)

fig.subplots_adjust(wspace=.25, hspace=.25)
    
for j, ax in enumerate(axs.flat):
    if j % 4 == 0:
        ax.set_ylabel('Inflammation \n regulation', fontsize = 12)
    if j >= 4*n_virs - 4:
        ax.set_xlabel('Antigenic \n regulation', fontsize = 12)
```
