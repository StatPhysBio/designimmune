```python
# Create additional variables
module_reg = {'$N \longrightarrow N^*$': [r for r in Na_reg if np.std(infection_scenarios[0][r].values) > 0], 
              '$N^* \longrightarrow E$': [r for r in NE_reg if np.std(infection_scenarios[0][r].values) > 0],  
              '$E \longrightarrow M$': [r for r in EM_reg if np.std(infection_scenarios[0][r].values) > 0], 
              '$E \longrightarrow E + E$': [r for r in EE_reg if np.std(infection_scenarios[0][r].values) > 0]}
module_reg_labels = {'$N \longrightarrow N^*$': [r for i, r in enumerate(param_names[-16:-12]) if  Na_reg[i] in module_reg['$N \longrightarrow N^*$']], 
                     '$N^* \longrightarrow E$': [r for i, r in enumerate(param_names[-12:-8]) if  NE_reg[i] in module_reg['$N^* \longrightarrow E$']],
                     '$E \longrightarrow M$': [r for i, r in enumerate(param_names[-8:-4]) if  EM_reg[i] in module_reg['$E \longrightarrow M$']],
                     '$E \longrightarrow E + E$': [r for i, r in enumerate(param_names[-4:]) if  EE_reg[i] in module_reg['$E \longrightarrow E + E$']]}

#variables = list(np.concatenate([module_reg[key] for key in module_reg]))
signal_classes = [r for i, r in enumerate(["Ag.", "Inflam.", "Anti-inflam.", "Baseline"]) if np.std(infection_scenarios[0][[Na_reg[i]] + [NE_reg[i]] + [EM_reg[i]] + [EE_reg[i]]].values) > 0]
stim_vars = ['stim_pI', 'stim_pH', 'stim_pP']

if "full" in comment or "sparse" in comment:
    reg = [r for r in Na_reg + NE_reg + EM_reg + EE_reg if np.std(infection_scenarios[0][r].values) > 0]
    reg_label = [r for i, r in enumerate(param_names[-16:]) if  (Na_reg + NE_reg + EM_reg + EE_reg)[i] in reg] 
    module_labels = ['$N \longrightarrow N^*$', '$N^* \longrightarrow E$', '$E \longrightarrow M$', '$E \longrightarrow E + E$']
    reg_or = [Na_reg[0:2]] + [NE_reg[0:2]] + [EM_reg[0:2]] + [EE_reg[0:2]]
    reg_and = [Na_reg[2]] + [NE_reg[2]] + [EM_reg[2]] + [EE_reg[2]]
    reg_and_label = [param_names[-14]] + [param_names[-10]] + [param_names[-6]] + [param_names[-2]]
    reg_bl = [Na_reg[3]] + [NE_reg[3]] + [EM_reg[3]] + [EE_reg[3]]
    reg_bl_label = [param_names[-13]] + [param_names[-9]] + [param_names[-5]] + [param_names[-1]]
else:
    reg = [r for r in ((Na_reg if "Na" in comment else []) + (NE_reg if "NE" in comment else []) + (EM_reg if "EM" in comment else []) + (EE_reg if "EE" in comment else [])) if np.std(infection_scenarios[0][r].values) > 0]
    reg_label = [r for i, r in enumerate(param_names[-16:])  if  (Na_reg + NE_reg + EM_reg + EE_reg)[i] in reg] 
    module_labels = (['$N \longrightarrow N^*$'] if "Na" in comment else []) + (['$N^* \longrightarrow E$'] if "NE" in comment else []) + (['$E \longrightarrow M$'] if "EM" in comment else []) + (['$E \longrightarrow E + E$'] if "EE" in comment else [])
    reg_or = (Na_reg[0:2] if "Na" in comment else []) + (NE_reg[0:2] if "NE" in comment else []) + (EM_reg[0:2] if "EM" in comment else []) + (EE_reg[0:2] if "EE" in comment else [])
    reg_and = ([Na_reg[2]] if "Na" in comment else []) + ([NE_reg[2]] if "NE" in comment else []) + ([EM_reg[2]] if "EM" in comment else []) + ([EE_reg[2]] if "EE" in comment else [])
    reg_and_label = ([param_names[-14]] if "Na" in comment else []) + ([param_names[-10]] if "NE" in comment else []) + ([param_names[-6]] if "EM" in comment else []) + ([param_names[-2]] if "EE" in comment else [])
    reg_bl = ([Na_reg[3]] if "Na" in comment else []) + ([NE_reg[3]] if "NE" in comment else []) + ([EM_reg[3]] if "EM" in comment else []) + ([EE_reg[3]] if "EE" in comment else [])
    reg_bl_label = ([param_names[-13]] if "Na" in comment else []) + ([param_names[-9]] if "NE" in comment else []) + ([param_names[-5]] if "EM" in comment else []) + ([param_names[-1]] if "EE" in comment else [])

num_categories = len(perf_labels)
reg_markers = ['o', 's', 'v', 'X', '^']
num_modules = len(module_labels)
num_psis = len(Na_reg)
```

```python
# Plot results from regressing performance by regulatory weights
for module_index in np.arange(0, num_modules):
    reg_choice_label = module_labels[module_index]
    reg_choice = module_reg[reg_choice_label]
    
    x_varname= key_var
    fig_c, axs = plt.subplots(num_categories, len(reg_choice), 
                              figsize=(len(reg_choice)*len(reg_choice), np.maximum(3.5, len(reg_choice))*num_categories), sharex='row', sharey= False)
    
    for i, var in enumerate(perf_vars):
        filter = impact_data.loc[(impact_data['perf_index'] == 0)*(impact_data['max_perf'] > 0.0)*(impact_data['harm_pI_noprotection'].values > 0.0)].copy()
        data = impact_data.loc[(impact_data['perf_index'] == i)*(impact_data['K_EH'] == K_EH)].copy()
        data.sort_values(by=x_varname, inplace=True)
        
        vir_data_mean = data.loc[[(u in filter[x_varname].values) for u in data[x_varname].values]].copy()

        for j, p in enumerate(signal_classes):
            y_varname = reg_choice[j]
            y_label = p+" impact, \n"+'$w[$'+module_reg_labels[reg_choice_label][j]+'$]$'

            x, y, z, err = vir_data_mean[x_varname], vir_data_mean[y_varname]/(vir_data_mean[stim_vars[j]] if "Base" not in p else 1.0), vir_data_mean['y_range'], vir_data_mean['sigma_'+y_varname]

            im = axs[i, j].scatter(x, y,
                              marker =reg_markers[0], s = 20, c = z, cmap='coolwarm_r', edgecolors = 'None', 
                                   norm = mpl.colors.Normalize())
            # mpl.colors.SymLogNorm(linthresh=0.01, linscale=1) if 'Memory' in perf_labels[i] or 'magnitude' in perf_labels[i] else 
            
            # Evaluate the spline and interpolation
            if len(x) >= 20:
                y_sg = savgol_filter(y, len(x) // 2, np.minimum(len(x) // 2, 3))
                axs[i, j].errorbar(x, y_sg, yerr = None, errorevery = 20, zorder = 1, label='Savitzky-Golay', color = 'k', linewidth = 2.0)
                
            #axs[i, j].set_title("Regulatory effect", fontsize = 10.0)
            axs[i, j].axhline(y = 0.0, color = 'k', linestyle = '--')
            axs[i, j].minorticks_off()
            axs[i, j].tick_params(axis='both', which='major', labelsize=8)
            axs[i, j].yaxis.set_tick_params(labelbottom=True)
            axs[i, j].set_box_aspect(1)
            axs[i, j].locator_params(tight=True, nbins=4)
            axs[i, j].set_xlabel(key_var_label, fontsize = 11)
            axs[i, j].set_ylabel(y_label, fontsize = 11)

            cb = fig_c.colorbar(im, ax=axs[i, j], orientation='vertical', shrink = 0.6, pad = 0.05)
            cb.set_label(perf_labels[i], fontsize = 10)
            cb.ax.tick_params(labelsize=8)
            
    fig_c.subplots_adjust(wspace=.75, hspace=0.5)

    plt.savefig('_figs/'+comment+'-'+reg_choice_label+'-regulation_impact_on_performance'+'.png', dpi=150, bbox_inches="tight")
```

### (b) Impact of cell fate modules on the immune response

```python
# Impactfulness of networks
clustered_mean_df['antigenicity_over_harm'] = antigenicity_over_harm(clustered_mean_df)

x_varname= key_var
fig_j, axs = plt.subplots(num_categories, num_modules, figsize=(num_modules*np.maximum(3.5, num_modules), num_categories*np.maximum(3.5, num_modules)), sharex='row', sharey='row')

for y in np.arange(0, num_modules):
    reg_choice_label = module_labels[y]
    reg_choice = module_reg[reg_choice_label]
    vir_data_mean = clustered_mean_df[reg + vir_vars + [x_varname] + perf_vars].groupby(vir_vars + [x_varname] + reg_choice, as_index=False).mean()
    data_std = vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).std()
    data_mean = np.abs(vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).mean())
    data_max = vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).max() - data_mean
    data_min = data_mean - vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).max()
    
    
    for i, category in enumerate(perf_labels):
        y_varname = perf_vars[i]

        im = axs[i, y].scatter(data_std[x_varname], data_std[y_varname]/data_mean[y_varname], 
                              marker ='s', s = 50, c = data_mean['b_I']*data_mean['S_0'] - data_mean['d_I'], cmap='coolwarm_r', edgecolors = 'None', norm=mpl.colors.Normalize())
        axs[i, y].scatter(data_mean[x_varname], np.maximum(data_max[y_varname],data_min[y_varname])/data_mean[y_varname], marker = '_', 
                            s = 50, c = data_mean['b_I']*data_mean['S_0'] - data_mean['d_I'], cmap='coolwarm_r', linewidths=2.5, label = r'Largest observed effect' if (i == 0)*(y == num_modules -1) else '')
        #axs[i, y].semilogx()
        axs[i, y].semilogy()
        axs[i, y].yaxis.set_tick_params(labelbottom=True)
        
        axs[i, y].minorticks_off()
        axs[i, y].tick_params(axis='both', which='major', labelsize=8)
        axs[i, y].set_box_aspect(1)
        axs[i, y].set_title(perf_labels[i]+'\n'+ module_labels[y], fontsize = 10.0)
        cb = fig_j.colorbar(im, ax=axs[i, y], orientation='vertical', shrink = 0.75, pad = 0.05)
        cb.set_label("Reproduction rate, \n"+r"$b_I\times S(0) - d_I$", fontsize = 10)
        cb.ax.tick_params(labelsize=8)
        

axs[0,-1].legend(loc='center right', bbox_to_anchor=(3.5, 0.5), fontsize = 12)
fig_j.subplots_adjust(wspace=.65, hspace=.85)

for k, ax in enumerate(axs.flat):
    ax.set_xlabel(key_var_label, fontsize = 12)
    ax.set_ylabel(r"$\frac{\text{Standard deviation}}{\text{Mean}}$", fontsize = 12)
        
plt.savefig('_figs/'+comment+'-modularity_by_performance_w_std'+'.png', dpi=150, bbox_inches="tight")
```

### (d) Dependence of performance on modules and architectures

```python
# Define utility and constraint functions
from scipy.optimize import minimize, Bounds, NonlinearConstraint, basinhopping, brute, fmin
opt_vars = ['psi_I', 'psi_H', 'psi_P', reg_bl[0], reg_bl[1], reg_bl[3]]
opt_vars_labels = signal_classes[0:3] + [reg_bl_label[0], reg_bl_label[1]+", "+reg_bl_label[2], reg_bl_label[3]]

def outcome(x, antigenicity_over_harm = 0.0, K_EH = K_EH, perf_index = 0, type = "average"):

    vars = np.array([x[0], x[1], x[2], x[3], # Na
           x[0], x[1], x[2], x[4], # NE
           -x[0], -x[1], -x[2], -x[4], # EM
           x[0], x[1], x[2], x[5]])  # EE
    xs = vars.reshape(-1,len(vars))
    
    if "average" in type:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data['K_EH'] == K_EH)*(impact_data['harm_pI_noprotection'].values > 0.0)]
    elif "single" in type:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data.antigenicity_over_harm == antigenicity_over_harm)*(impact_data['K_EH'] == K_EH)]
    elif "auto" in type:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data.b_I == 0.0)*(impact_data['K_EH'] == K_EH)*(impact_data['K_IE'] == S_0)] # *(impact_data['K_IE'] == S_0)*(impact_data['d_I'] > d_S)
 
    out = np.array([sigmoid(xs, *p).item() for p in input[['y_range', 'y_min'] + reg + [('intercept_' + j) for j in modules]].values])
    return np.mean(out)

def compute_grad(x, antigenicity_over_harm = 0.0, K_EH = K_EH, perf_index = 0, type = "average", h = 1e-6):
    out = np.zeros(len(x))
    for i in np.arange(len(x)):
        dx = np.zeros(len(x))
        dx[i] += h
        f = np.hstack([outcome(x - dx , 0, K_EH, perf_index, type), outcome(x, 0, K_EH, perf_index, type), outcome(x + dx , 0, K_EH, perf_index, type)])
        grad = np.gradient(f, h)
        out[i] += grad[1]

    return out

def moo_2d(x, antigenicity_over_harm = 0.0, K_EH = K_EH, perf_index_1 = 0, perf_index_2 = 0, type_1 = "average", type_2 = "average"):
    out = [outcome(x, antigenicity_over_harm, K_EH, perf_index_1, type_1), outcome(x, antigenicity_over_harm, K_EH, perf_index_2, type_2)]
    grad1 = compute_grad(x, antigenicity_over_harm, K_EH, perf_index_1, type_1).tolist()
    grad2 = compute_grad(x, antigenicity_over_harm, K_EH, perf_index_2, type_2).tolist()
    
    return out + grad1 + grad2

def utility(x, antigenicity_over_harm = 0.0, K_EH = K_EH, lam = 1, type = "average_prot_tox"):
    if "prot_tox" in type:
        out = outcome(x, antigenicity_over_harm, K_EH, 0, type = type) - lam*outcome(x, antigenicity_over_harm, K_EH, 1, type = type)
    elif "prot_mem" in type:
        out = outcome(x, antigenicity_over_harm, K_EH, 0, type = type) - outcome(x, antigenicity_over_harm, K_EH, 1, type = type) + lam*outcome(x, antigenicity_over_harm, K_EH, 3, type = type)
    elif "prot_auto" in type:
        out = outcome(x, antigenicity_over_harm, K_EH, 0, type = type) - outcome(x, antigenicity_over_harm, K_EH, 1, type = type) - lam*outcome(x, antigenicity_over_harm, K_EH, 1, type = "auto")*(np.exp(outcome(x, antigenicity_over_harm, K_EH, 2, type = type)) - 1)*outcome(x, antigenicity_over_harm, K_EH, 3, type = type)
    elif "auto" in type:
        out = lam*(outcome(x, antigenicity_over_harm, K_EH, 0, type = type) - outcome(x, antigenicity_over_harm, K_EH, 1, type = type)) - outcome(x, antigenicity_over_harm, K_EH, 1, type = "auto")*(np.exp(outcome(x, antigenicity_over_harm, K_EH, 2, type = type)) - 1)*outcome(x, antigenicity_over_harm, K_EH, 3, type = type)

    return -out

def constraint1(x, bound = psi_max):
    return np.sum(np.abs(x[0:3])) - bound

# init bounds:
init_point = -np.array([0.0, 0.0, 0.0] + (np.mean(impact_data.loc[impact_data.perf_index == 0,[ 'intercept_'+r for r in [modules[0], modules[1], modules[3]] ]], axis = 0).values \
                       /np.mean(impact_data.loc[impact_data.perf_index == 0, [ 'F0_'+r for r in [modules[0], modules[1], modules[3]] ]], axis = 0).values).tolist()) # np.zeros(len(opt_vars)) 
lower_bounds = np.minimum(init_point + np.array(2*[-1*psi_max] + [-1*psi_max] + (num_modules -1)*[-1*F0_max]), np.array(2*[-1*psi_max] + [-1*psi_max] + (num_modules -1)*[-1*F0_max]))
upper_bounds = np.maximum(init_point + np.array(2*[1*psi_max] + [1*psi_max] + (num_modules -1)*[1*F0_max]), np.array(2*[1*psi_max] + [1*psi_max] + (num_modules -1)*[1*F0_max]))

class BoundStep:
   def __init__(self, stepsize = 2.0*max(np.abs(upper_bounds))):
       self.stepsize = stepsize
       self.rng = np.random.default_rng()
   def __call__(self, x):
       s = self.stepsize

       x += self.rng.uniform(-s, s, x.shape)
       return np.clip(x, a_min = lower_bounds, 
                      a_max = upper_bounds) #np.minimum(np.maximum(x, -3*psi_max), 3*psi_max)

boundstep = BoundStep()

class regBounds(object):
    ''' 
    bounds class to make sure your variable is with in the inspected bounds
    '''
    def __init__(self, xmin, xmax):
        self.xmax = np.array(xmax)
        self.xmin = np.array(xmin)

    def __call__(self, **kwargs):
        x = kwargs["x_new"]
        tmax = bool(np.all(x <= self.xmax))
        tmin = bool(np.all(x >= self.xmin))
        return tmax and tmin

acceptbounds = regBounds(lower_bounds, upper_bounds)

# define constraints
# const1 = NonlinearConstraint(constraint1, -np.inf, 0)
bounds = [(low, high) for low, high in zip(lower_bounds, upper_bounds)]
num_cpu = 150
tol = 1e-9
```

### (e) Sensitivity vs Specificity Trade-off

```python
# Regulation of networks versus performance
fig_h, axs = plt.subplots(len(perf_vars), len(reg_and + reg_bl), figsize=(len(reg_and + reg_bl)**2, len(perf_vars)**2), sharex='row', sharey='row')

# calculate sensitivity
utility_df = clustered_mean_df.groupby(reg, as_index=False).mean()
utility_df.sort_values(by=reg, inplace=True)

# calculate specificity
risk_df = clustered_mean_df.groupby(reg, as_index=False).std()
risk_df.sort_values(by=reg, inplace=True)

for p, var in enumerate(perf_vars):
    indices = [0,1,4,5,8,9,12,13]
    x = utility_df[var]
    y = risk_df[var]/np.abs(utility_df[var])
    
    for j, reg_var in enumerate(reg_and + reg_bl):
        z = utility_df[reg[indices[j]]]
        im = axs[p, j].scatter(x, y, 
                                      marker ='o', s = 10, c = z, cmap='coolwarm_r', edgecolors = 'None', alpha = 0.5,
                              norm = mpl.colors.Normalize(vmin= -psi_max/2, vmax= psi_max/2))
        axs[p, j].set_title((perf_labels[p] +' \n' if j == 0 else '') + reg_label[indices[j]], fontsize = 12.0)
        axs[p, j].set_yscale("symlog")
        axs[p, j].minorticks_off()
        axs[p, j].yaxis.set_tick_params(labelbottom=True)
        axs[p, j].set_box_aspect(1)
        
        cb = fig_h.colorbar(im, ax=axs[p, j], orientation='vertical', shrink = 0.75, pad = 0.05)
        cb.set_label("Regulatory parameter", fontsize = 10)
            
fig_h.subplots_adjust(wspace=.5, hspace=0.25)

for k, ax in enumerate(axs.flat):
    ax.set_ylabel("Coef.\n of variation", fontsize = 12)
    ax.set_xlabel("Average performance", fontsize = 12)
        
plt.savefig('_figs/'+comment+'-'+'_utility_v_risk_by_reg_sctr-'+var+'.png', dpi=150, bbox_inches="tight")
```
