### (d) Dependence of performance on modules and architectures

```python
# Define utility and constraint functions
from scipy.optimize import minimize, Bounds, NonlinearConstraint, basinhopping, brute, fmin

opt_vars = ['psi_I', 'psi_HI', 'psi_HE', reg_bl[0], reg_bl[1], reg_bl[2], reg_bl[3]]
opt_vars_labels = [signal_classes[0]+r", $\psi_{E}^{(Ag)}$", 
                   signal_classes[1]+r", $\psi_{E}^{(H_I)}$", 
                   signal_classes[2]+r", $\psi_{E}^{(H_E)}$",
                   module_labels[0]+r" base., $L_{0,N \to N^*}$", 
                   module_labels[1]+r" base., $L_{0,N^* \to E}$",
                   module_labels[2]+r" base., $L_{0,E \to M}$",
                   module_labels[3]+r" base., $L_{0,E \to 2E}$"]
linestyles = ['-','--','-','-.',':','--', '-']

def outcome(x, antigenicity_over_harm = 0.0, N_0 = N_0, perf_index = 0, type = "average", constraint = True):

    # Enforce regulatory constraints
    # There are still some questions about how the efector to memory transition should be dealt with.
    if constraint == True:
        vars = np.array([x[0], x[1], x[2], x[3], # Na
               x[0], x[1], x[2], x[4], # NE
               -x[0], -x[1], -x[2], x[5], # EM
               x[0], x[1], x[2], x[6]])  # EE
    else:
        vars = x
    
    xs = vars.reshape(-1,len(vars))
    
    if "average" in type:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data['b_I'].values > 0.0)*(impact_data.I_0 == I_0)*(impact_data.N_0 == N_0)]
    elif "single" in type:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data.antigenicity_over_harm == antigenicity_over_harm)*(impact_data.N_0 == N_0)]
    elif "auto" in type:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data.b_I == 0.0)*(impact_data['K_IE'] == K_SE)*(impact_data['d_I'] == d_S)*(impact_data['N_0'] == N_0)]
    elif "cancer" in type:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data.I_0 > I_0)]

    if perf_index == 2 or perf_index == 3:
        out = np.mean(np.array([10**a - 1 for a in np.array([sigmoid(xs, *p).item() for p in input[['y_range', 'y_min'] + reg + [('intercept_' + j) for j in modules]].values])]))
    else:
        out = np.mean(np.array([sigmoid(xs, *p).item() for p in input[['y_range', 'y_min'] + reg + [('intercept_' + j) for j in modules]].values]))
        
    return out

def compute_grad(x, antigenicity_over_harm = 0.0, N_0 = N_0, perf_index = 0, type = "average", h = 1e-9):
    out = np.zeros(len(x))
    for i in np.arange(len(x)):
        dx = np.zeros(len(x))
        dx[i] += h
        if perf_index == "util":
            f = np.hstack([outcome(x - dx , antigenicity_over_harm, N_0, 0, type) - outcome(x - dx , antigenicity_over_harm, N_0, 1, type), 
                           outcome(x, antigenicity_over_harm, N_0, 0, type) - outcome(x, antigenicity_over_harm, N_0, 1, type), 
                           outcome(x + dx , antigenicity_over_harm, N_0, 0, type) - outcome(x + dx , antigenicity_over_harm, N_0, 1, type)])
        elif perf_index == "auto":
            f = np.hstack([outcome(x - dx, antigenicity_over_harm, N_0, 1, type = "auto"),
                           outcome(x, antigenicity_over_harm, N_0, 1, type = "auto"),
                           outcome(x + dx, antigenicity_over_harm, N_0, 1, type = "auto")])
        elif perf_index == "mem":
            f = np.hstack([outcome(x - dx, antigenicity_over_harm, N_0, 2, type = type) + outcome(x - dx, antigenicity_over_harm, N_0, 3, type = type)*outcome(x - dx, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25),
                           outcome(x, antigenicity_over_harm, N_0, 2, type = type) + outcome(x, antigenicity_over_harm, N_0, 3, type = type)*outcome(x, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25),
                           outcome(x + dx, antigenicity_over_harm, N_0, 2, type = type) + outcome(x + dx, antigenicity_over_harm, N_0, 3, type = type)*outcome(x + dx, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25)])
        elif perf_index == "dur_emem":
            f = np.hstack([outcome(x - dx, antigenicity_over_harm, N_0, 3, type = type)*outcome(x - dx, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25),
                           outcome(x, antigenicity_over_harm, N_0, 3, type = type)*outcome(x, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25),
                           outcome(x + dx, antigenicity_over_harm, N_0, 3, type = type)*outcome(x + dx, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25)])
        else:
            f = np.hstack([outcome(x - dx , antigenicity_over_harm, N_0, perf_index, type), 
                           outcome(x, antigenicity_over_harm, N_0, perf_index, type), 
                           outcome(x + dx , antigenicity_over_harm, N_0, perf_index, type)])
            
        grad = np.gradient(f, h)
        out[i] += grad[1]

    return out

def moo_2d(x, antigenicity_over_harm = 0.0, N_0 = N_0, perf_index_1 = 0, perf_index_2 = 0, type_1 = "average", type_2 = "average"):
    if perf_index_1 == "util":
        out1 = outcome(x, antigenicity_over_harm, N_0, 0, type_1) - outcome(x, antigenicity_over_harm, N_0, 1, type_1)
    elif perf_index_1 == "auto":
        out1 = outcome(x, antigenicity_over_harm, N_0, 1, type = "auto")
    elif perf_index_1 == "mem":
        out1 = outcome(x, antigenicity_over_harm, N_0, 2, type = type_1) + outcome(x, antigenicity_over_harm, N_0, 3, type = type_1)*outcome(x, antigenicity_over_harm, N_0, 4, type = type_1)/(t_long_M/365.25)
    elif perf_index_1 == "dur_emem":
        out1 = outcome(x, antigenicity_over_harm, N_0, 3, type = type_1)*outcome(x, antigenicity_over_harm, N_0, 4, type = type_1)/(t_long_M/365.25)
    else:
        out1 = outcome(x, antigenicity_over_harm, N_0, perf_index_1, type_1)

    if perf_index_2 == "util":
        out2 = outcome(x, antigenicity_over_harm, N_0, 0, type_2) - outcome(x, antigenicity_over_harm, N_0, 1, type_2)
    elif perf_index_2 == "auto":
        out2 = outcome(x, antigenicity_over_harm, N_0, 1, type = "auto")
    elif perf_index_2 == "mem":
        out2 = outcome(x, antigenicity_over_harm, N_0, 2, type = type_2) + outcome(x, antigenicity_over_harm, N_0, 3, type = type_2)*outcome(x, antigenicity_over_harm, N_0, 4, type = type_2)/(t_long_M/365.25)
    elif perf_index_2 == "dur_emem":
        out2 = outcome(x, antigenicity_over_harm, N_0, 3, type = type_2)*outcome(x, antigenicity_over_harm, N_0, 4, type = type_2)/(t_long_M/365.25)
    else:
        out2 = outcome(x, antigenicity_over_harm, N_0, perf_index_2, type_2)
    
    out = [out1, out2]
    grad1 = compute_grad(x, antigenicity_over_harm, N_0, perf_index_1, type_1).tolist()
    grad2 = compute_grad(x, antigenicity_over_harm, N_0, perf_index_2, type_2).tolist()
    
    return out + grad1 + grad2

def sample_3d_simplex(weights = [1,1,1], pnts = 100):
    """Samples points uniformly from the 3D simplex."""
    x = np.random.dirichlet(weights, pnts)
    
    return x

def utility(x, antigenicity_over_harm = 0.0, N_0 = N_0, lam = [1,0,0], type = "average_clear_tox", max_val = [1.0,1.0,1.0]):
    if "util_cmem_ememdur" in type:
        out = lam[2]*(outcome(x, antigenicity_over_harm, N_0, 0, type = type) - outcome(x, antigenicity_over_harm, N_0, 1, type = type))/max_val[2] + lam[1]*outcome(x, antigenicity_over_harm, N_0, 2, type = type)/max_val[1] + lam[0]*outcome(x, antigenicity_over_harm, N_0, 4, type = type)/max_val[0]
    elif "clear_tox" in type:
        out = lam[1]*outcome(x, antigenicity_over_harm, N_0, 0, type = type) - lam[0]*outcome(x, antigenicity_over_harm, N_0, 1, type = type)
    elif "util_mem" in type:
        out = lam[1]*(outcome(x, antigenicity_over_harm, N_0, 0, type = type) - outcome(x, antigenicity_over_harm, N_0, 1, type = type)) + lam[0]*(outcome(x, antigenicity_over_harm, N_0, 2, type = type) + outcome(x, antigenicity_over_harm, N_0, 3, type = type)*outcome(x, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25))
    elif "util_cmem" in type:
        out = lam[1]*(outcome(x, antigenicity_over_harm, N_0, 0, type = type) - outcome(x, antigenicity_over_harm, N_0, 1, type = type)) + lam[0]*outcome(x, antigenicity_over_harm, N_0, 2, type = type)
    elif "util_dur_emem" in type:
        out = lam[1]*(outcome(x, antigenicity_over_harm, N_0, 0, type = type)/max_val[1] - outcome(x, antigenicity_over_harm, N_0, 1, type = type)) + lam[0]*outcome(x, antigenicity_over_harm, N_0, 3, type = type)*outcome(x, antigenicity_over_harm, N_0, 4, type = type)/(t_long_M/365.25)/max_val[0]
    elif "clear_cmem" in type:
        out = lam[1]*outcome(x, antigenicity_over_harm, N_0, 0, type = type) + lam[0]*outcome(x, antigenicity_over_harm, N_0, 2, type = type)
    elif "cmem_emem" in type:
        out = lam[1]*outcome(x, antigenicity_over_harm, N_0, 2, type = type)/max_val[1] + lam[0]*outcome(x, antigenicity_over_harm, N_0, 3, type = type)/max_val[0]
    elif "clear_auto" in type:
        out = lam[1]*outcome(x, antigenicity_over_harm, N_0, 0, type = type) - lam[0]*outcome(x, antigenicity_over_harm, N_0, 1, type = "auto")
    elif "util_auto" in type:
        out = lam[1]*(outcome(x, antigenicity_over_harm, N_0, 0, type = type) - outcome(x, antigenicity_over_harm, N_0, 1, type = type)) - lam[0]*outcome(x, antigenicity_over_harm, N_0, 1, type = "auto")
    elif "emem_auto" in type:
        out = lam[1]*outcome(x, antigenicity_over_harm, N_0, 3, type = type)/max_val[1] - lam[0]*outcome(x, antigenicity_over_harm, N_0, 1, type = "auto")/max_val[0]
    elif "emem_ememdur" in type:
        out = lam[1]*outcome(x, antigenicity_over_harm, N_0, 3, type = type)/max_val[1] + lam[0]*outcome(x, antigenicity_over_harm, N_0, 4, type = type)/max_val[0]
    elif "cmem_auto" in type:
        out = lam[1]*outcome(x, antigenicity_over_harm, N_0, 2, type = type) - lam[0]*outcome(x, antigenicity_over_harm, N_0, 1, type = "auto")
    elif "util_cancer" in type:
        out = lam[1]*(outcome(x, antigenicity_over_harm, N_0, 0, type = type) - outcome(x, antigenicity_over_harm, N_0, 1, type = type)) + lam[0]*(outcome(x, antigenicity_over_harm, N_0, 0, type = "cancer") - outcome(x, antigenicity_over_harm, N_0, 1, type = "cancer"))

    return -out

# init bounds:
init_point = np.zeros(len(opt_vars))
lower_bounds = np.minimum(init_point + np.array(3*[-2] + num_modules*[-1*10]), np.array(3*[-2] + num_modules*[-1*10]))
upper_bounds = np.maximum(init_point + np.array(3*[2] + num_modules*[1*10]), np.array(3*[2] + num_modules*[1*10]))

class BoundStep:
   def __init__(self, stepsize = 2.0*max(np.abs(upper_bounds))):
       self.stepsize = stepsize
       self.rng = np.random.default_rng()
   def __call__(self, x):
       s = self.stepsize

       x += self.rng.uniform(-s, s, x.shape)
       return np.clip(x, a_min = lower_bounds, 
                      a_max = upper_bounds) #np.minimum(np.maximum(x, -3*psi_max), 3*psi_max)

boundstep = BoundStep()

class regBounds(object):
    ''' 
    bounds class to make sure your variable is with in the inspected bounds
    '''
    def __init__(self, xmin, xmax):
        self.xmax = np.array(xmax)
        self.xmin = np.array(xmin)

    def __call__(self, **kwargs):
        x = kwargs["x_new"]
        tmax = bool(np.all(x <= self.xmax))
        tmin = bool(np.all(x >= self.xmin))
        return tmax and tmin

acceptbounds = regBounds(lower_bounds, upper_bounds)

# define constraints
bounds = [(low, high) for low, high in zip(lower_bounds, upper_bounds)]
num_cpu = 151
tol = 1e-6

# Define auto-reactivity slope
regress_auto = LinearRegression().fit(np.log10(1 + impact_data.loc[(impact_data.perf_index == 1)*(impact_data.b_I == 0.0)*(impact_data['K_IE'] == K_SE)*(impact_data['d_I'] == d_S)].N_0/N_0).values.reshape(-1, 1),
                            impact_data.loc[(impact_data.perf_index == 1)*(impact_data.b_I == 0.0)*(impact_data['K_IE'] == K_SE)*(impact_data['d_I'] == d_S)].y_range.values.reshape(-1, 1))
auto_coeff, auto_intercept = regress_auto.coef_.item(), regress_auto.intercept_.item()
print(auto_coeff, auto_intercept)
```

```python
# Code from: https://matplotlib.org/stable/gallery/images_contours_and_fields/image_annotated_heatmap.html
def heatmap(data, row_labels, col_labels, ax=None,
            cbar_kw=None, cbarlabel="", vmin = -2, vmax = 2, **kwargs):
    """
    Create a heatmap from a numpy array and two lists of labels.

    Parameters
    ----------
    data
        A 2D numpy array of shape (M, N).
    row_labels
        A list or array of length M with the labels for the rows.
    col_labels
        A list or array of length N with the labels for the columns.
    ax
        A `matplotlib.axes.Axes` instance to which the heatmap is plotted.  If
        not provided, use current Axes or create a new one.  Optional.
    cbar_kw
        A dictionary with arguments to `matplotlib.Figure.colorbar`.  Optional.
    cbarlabel
        The label for the colorbar.  Optional.
    **kwargs
        All other arguments are forwarded to `imshow`.
    """

    if ax is None:
        ax = plt.gca()

    if cbar_kw is None:
        cbar_kw = {}

    # Plot the heatmap
    norm = plt.cm.colors.Normalize(vmin=vmin, vmax=vmax)
    im = ax.imshow(data, norm = norm, **kwargs)

    # Create colorbar
    cbar = ax.figure.colorbar(im, ax=ax, **cbar_kw)
    cbar.ax.set_ylabel(cbarlabel, rotation=-90, va="bottom")

    # Show all ticks and label them with the respective list entries.
    ax.set_xticks(np.arange(data.shape[1]), labels=col_labels)
    ax.set_yticks(np.arange(data.shape[0]), labels=row_labels)

    # Let the horizontal axes labeling appear on top.
    ax.tick_params(top=True, bottom=False,
                   labeltop=True, labelbottom=False)

    # Rotate the tick labels and set their alignment.
    plt.setp(ax.get_xticklabels(), rotation=-30, ha="right",
             rotation_mode="anchor")

    # Turn spines off and create white grid.
    ax.spines[:].set_visible(False)

    ax.set_xticks(np.arange(data.shape[1]+1)-.5, minor=True)
    ax.set_yticks(np.arange(data.shape[0]+1)-.5, minor=True)
    ax.grid(which="minor", color="w", linestyle='-', linewidth=3)
    ax.tick_params(which="minor", bottom=False, left=False)

    return im, cbar


def annotate_heatmap(im, data=None, valfmt="{x:.2f}",
                     textcolors=("black", "white"),
                     threshold=None, **textkw):
    """
    A function to annotate a heatmap.

    Parameters
    ----------
    im
        The AxesImage to be labeled.
    data
        Data used to annotate.  If None, the image's data is used.  Optional.
    valfmt
        The format of the annotations inside the heatmap.  This should either
        use the string format method, e.g. "$ {x:.2f}", or be a
        `matplotlib.ticker.Formatter`.  Optional.
    textcolors
        A pair of colors.  The first is used for values below a threshold,
        the second for those above.  Optional.
    threshold
        Value in data units according to which the colors from textcolors are
        applied.  If None (the default) uses the middle of the colormap as
        separation.  Optional.
    **kwargs
        All other arguments are forwarded to each call to `text` used to create
        the text labels.
    """

    if not isinstance(data, (list, np.ndarray)):
        data = im.get_array()

    # Normalize the threshold to the images color range.
    if threshold is not None:
        threshold = im.norm(threshold)
    else:
        threshold = im.norm(data.max())/2.

    # Set default alignment to center, but allow it to be
    # overwritten by textkw.
    kw = dict(horizontalalignment="center",
              verticalalignment="center")
    kw.update(textkw)

    # Get the formatter in case a string is supplied
    if isinstance(valfmt, str):
        valfmt = mpl.ticker.StrMethodFormatter(valfmt)

    # Loop over the data and create a `Text` for each "pixel".
    # Change the text's color depending on the data.
    texts = []
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            kw.update(color=textcolors[int(im.norm(data[i, j]) > threshold)])
            text = im.axes.text(j, i, valfmt(data[i, j], None), **kw)
            texts.append(text)

    return texts
```

```python
# Regulation of highest ranking networks
import matplotlib.patches as mpatches
plots = [[vir_vars[1], vir_vars[0]]] #, [vir_vars[1], vir_vars[2]], [vir_vars[0], vir_vars[2]]]
plot_labels = [["Antigenicity, "+r"$\frac{S_{max}}{K_{I}}$", "Lethality, "+r"$\frac{d_{I}}{d_S}$"], 
               ["Antigenicity, "+r"$K_{I,E}^{-1}$", "Fecundity, "+r"$b_{I}$"], 
               ["Lethality, "+r"$d_{I}$", "Fecundity, "+r"$b_{I}$"]]
size = 2250/np.cbrt(n_virs)
infections = impact_data.loc[(impact_data['b_I'].values > 0.0)*(impact_data.I_0 == I_0)*(impact_data.perf_index == 0), 
vir_vars + [key_var, 'y_range']]
batch_size = max(int(len(infections)/num_cpu), 1)

fig_c, axs = plt.subplots(2, num_categories, 
                          figsize=(num_categories**2 - num_categories, 8), sharex=True, sharey=True)

for q, plot in enumerate(plots):
    
    x_varname, y_varname = plot[0], plot[1]
    
    for i, category in enumerate(perf_labels):
        derivative_data = pd.concat([infections.reset_index(drop=True), 
                                     pd.DataFrame(np.vstack(Parallel(n_jobs = np.minimum(num_cpu, len(infections)), 
                                         batch_size = batch_size)(delayed(moo_2d)(init_point, a[0], a[1], perf_index_1 = i, perf_index_2 = 0, type_1 = "single", type_2 = "single")
                                                                             for a in infections[[key_var, 'N_0']].values))[:, 2:2 + len(opt_vars)], columns = ["grad_"+ var for var in opt_vars]).reset_index(drop=True)],
                                    axis = 1)

        for j, p in enumerate(["Feedback:", "Baseline:"]):

            # compute dominant regulatory feedback signal or threshold state
            input = np.abs(derivative_data[["grad_"+ var for var in opt_vars[j*3:(j+1)*3 + j]]].values)/(np.sum(np.abs(derivative_data[["grad_"+ var for var in opt_vars[j*3:(j+1)*3 + j]]]), axis = 1).values)[:, np.newaxis]
            #input = data[opt_vars[j*3:(j+1)*3 + j]]/data[intercept_vars[j]]/np.sum(np.log(1 + np.exp(-data[intercept_vars])).values, axis = 1)
            output = np.argmax(input, axis = 1) 
            #np.array([ np.nonzero(a)[0].item() if np.nonzero(a)[0].size > 0 else -1 for a in input > 1/2 ])
            scatter_x = (1/derivative_data[x_varname] if x_varname == 'K_IE' else derivative_data[x_varname]).values
            scatter_y = (1/derivative_data[y_varname] if y_varname == 'K_IE' else derivative_data[y_varname]).values
            
            for g in np.unique(output):
                ix = np.where(output == g)  
                im = axs[j, i].scatter(S_0*scatter_x[ix], scatter_y[ix]/d_S,
                                   facecolors = cmap_colors[j*3 + g], alpha = np.minimum(1, 2*input[ix, g]) if g >=0 else 1.0,
                                   marker ='s', s = size, linewidth=0 if g >=0 else 0.25, edgecolors='k' if g < 0 else 'none',
                                   label = g) #, hatch=['/' if d == 0 else '.' for d in np.sum((input >= 1/2), axis = 1)[ix]])

            if x_varname == 'K_IE':
                axs[j, i].semilogx()
            if y_varname == 'K_IE':
                axs[j, i].semilogy()
                
            axs[j, i].minorticks_off()
            axs[j, i].set_box_aspect(1)
            axs[j, i].locator_params(axis='y', nbins=6)
            axs[j, i].tick_params(axis='both', which='major', labelsize=10)
            if j == 0:
                axs[j, i].set_title(perf_labels[i], fontsize = 16.0)

            # Create legend handles manually
            if i == num_categories - 1:
                legend_handles = []
                for key, color in enumerate(cmap_colors[j*3:(j+1)*3 + j]):
                    patch = mpatches.Patch(color=color, label=(opt_vars_labels[j*3:(j+1)*3 + j])[key]) # ["Mixed"] + 
                    legend_handles.append(patch)
                
                # Add the legend
                axs[j, i].legend(handles=legend_handles, bbox_to_anchor=(2.2, 1), fontsize=14, framealpha=0, ncol = 1)
            
    fig_c.subplots_adjust(wspace=.15, hspace=0.15)
    
    for k, ax in enumerate(axs.flat):
        if k >= num_categories:
            ax.set_xlabel(plot_labels[q][0], fontsize = 16)
        if k % num_categories == 0:
            ax.set_ylabel(plot_labels[q][1], fontsize = 16)

    plt.savefig('_figs/'+comment+'-regulatory_effect_by_performance_phase_dgram'+'.png', dpi=150, bbox_inches="tight")
```

#### Optimize $(1 - \lambda)\cdot[\text{Clearance}] - \lambda\cdot [\text{Toxicity}]$

```python
# Optimize average utlity
import warnings
warnings.filterwarnings('ignore') # suppress warnings 

lams = np.linspace(0, 1, num_cpu)
res_avg = Parallel(n_jobs = np.minimum(num_cpu, len(lams)), 
         batch_size = max(int(len(lams)/np.minimum(num_cpu, len(lams))), 1))(delayed(basinhopping)(utility, 
                                                                                                   x0 = init_point, 
                                                                                                   minimizer_kwargs = {"method": "L-BFGS-B", "args": (0.0, N_0, [l, 1-l, 0]), "bounds": bounds},
                                                                                                   T = 0.0,
                                                                                                   stepsize = 2*max(np.abs(upper_bounds)),
                                                                                                   take_step = boundstep,
                                                                                                   accept_test = acceptbounds,
                                                                                                   niter= 75)
                                                                             for l in lams)

# reshape data
avg_util_optimum_data = np.vstack([-u.fun for u in res_avg])
avg_util_psi_opts_data = np.vstack([u.x for u in res_avg])
plt.scatter(1-lams, avg_util_optimum_data)
plt.ylabel("Protection \n[$d_S\cdot S_{max}\cdot T_{sim}$]")
plt.xlabel("Clearance priority, $\lambda$")
```

```python
# Inspect pareto fronts
infections = impact_data.loc[(impact_data['b_I'].values > 0.0)*(impact_data.I_0 == I_0)*(impact_data.perf_index == 0), vir_vars + [key_var]]
run_scenarios = pd.DataFrame(np.vstack([a + [l] for a, l in list(itertools.product(infections.values.tolist(),
                                lams.tolist()))]), columns = vir_vars + [key_var, 'lambda'])

res_infections = Parallel(n_jobs = np.minimum(num_cpu, len(infections)), 
         batch_size = len(lams))(delayed(basinhopping)(utility, 
                                                               x0 = init_point,
                                                               minimizer_kwargs = {"method": "L-BFGS-B", "args": (a, N_0, [l, 1-l, 0], "single_clear_tox"), "tol": tol, "bounds": bounds},
                                                               T = 0.0,
                                                               stepsize = 2.0*max(np.abs(upper_bounds)),
                                                               take_step = boundstep,
                                                               accept_test = acceptbounds,
                                                               niter= 75)
                                         for a, l in zip(run_scenarios[key_var].values, run_scenarios['lambda'].values)
                                                    )

# reshape data
single_clear_tox_psi_opts_data = pd.concat([run_scenarios.reset_index(drop=True), 
                                           pd.DataFrame(np.vstack([u.x for u in res_infections]), columns = opt_vars).reset_index(drop=True)], 
                                          axis=1)
single_clear_tox_psi_opts_data["perf"] = np.vstack([-u.fun for u in res_infections])

plt.scatter(single_clear_tox_psi_opts_data[key_var], 
            single_clear_tox_psi_opts_data["perf"])
plt.ylabel("Protection \n[$d_S\cdot S_{max}\cdot T_{sim}$]")
plt.xlabel(key_var_label)
```

```python
# Optimize average utlity
import warnings
warnings.filterwarnings('ignore') # suppress warnings 

lams_mem = np.linspace(0, 1, num_cpu)
#np.concatenate([np.zeros(1), np.logspace(-2.6, -2.575, num_cpu - 2), np.ones(1)])
res_avg = Parallel(n_jobs = np.minimum(num_cpu, len(lams)), 
         batch_size = max(int(len(lams)/np.minimum(num_cpu, len(lams))), 1))(delayed(basinhopping)(utility, 
                                                                                                   x0 = init_point, 
                                                                                                   minimizer_kwargs = {"method": "L-BFGS-B", "args": (0.0, N_0, [l, 1-l, 0], "average_cmem_emem"),
                                                                                                                                                     #[np.amax(data_cmem_emem[:,1]), np.amax(data_cmem_emem[:,0]), 1.0]),
                                                                                                                       "tol": tol, "bounds": bounds},
                                                                                                   T = 0.0,
                                                                                                   stepsize = 2*max(np.abs(upper_bounds)),
                                                                                                   take_step = boundstep,
                                                                                                   accept_test = acceptbounds,
                                                                                                   niter= 75)
                                                                             for l in lams_mem)

# reshape data
avg_mem_optimum_data = np.vstack([-u.fun for u in res_avg])
avg_mem_psi_opts_data = np.vstack([u.x for u in res_avg])
plt.scatter(lams_mem, avg_mem_optimum_data)
plt.ylabel("Objective")
plt.xlabel("Expansion priority, $\lambda$")
```
