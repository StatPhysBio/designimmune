## 1. Performance-based evaluation of immune networks

```python
# Create additional variables
vir_vars = ['d_I', 'K_IE', 'b_I', 'K_EH', 'S_0']

module_reg = {'$N \longrightarrow N^*$': [r for r in Nact_reg if np.std(infection_scenarios[0][r].values) > 0], 
              '$N^* \longrightarrow M$': [r for r in NM_reg if np.std(infection_scenarios[0][r].values) > 0],  
              '$E \longrightarrow M$': [r for r in EM_reg if np.std(infection_scenarios[0][r].values) > 0], 
              '$E \longrightarrow E + E$': [r for r in Ediv_reg if np.std(infection_scenarios[0][r].values) > 0]}
module_reg_labels = {'$N \longrightarrow N^*$': [r for i, r in enumerate(param_names[-16:-12]) if  Nact_reg[i] in module_reg['$N \longrightarrow N^*$']], 
                     '$N^* \longrightarrow M$': [r for i, r in enumerate(param_names[-12:-8]) if  NM_reg[i] in module_reg['$N^* \longrightarrow M$']],
                     '$E \longrightarrow M$': [r for i, r in enumerate(param_names[-8:-4]) if  EM_reg[i] in module_reg['$E \longrightarrow M$']],
                     '$E \longrightarrow E + E$': [r for i, r in enumerate(param_names[-4:]) if  Ediv_reg[i] in module_reg['$E \longrightarrow E + E$']]}

#variables = list(np.concatenate([module_reg[key] for key in module_reg]))
signal_classes = [r for i, r in enumerate(["Ag.", "Inflam.", "Anti-inflam.", "Baseline"]) if np.std(infection_scenarios[0][[Nact_reg[i]] + [NM_reg[i]] + [EM_reg[i]] + [Ediv_reg[i]]].values) > 0]

if "full" in comment or "sparse" in comment:
    reg = [r for r in Nact_reg + NM_reg + EM_reg + Ediv_reg if np.std(infection_scenarios[0][r].values) > 0]
    reg_label = [r for i, r in enumerate(param_names[-16:]) if  (Nact_reg + NM_reg + EM_reg + Ediv_reg)[i] in reg] 
    modules = ['$N \longrightarrow N^*$', '$N^* \longrightarrow M$', '$E \longrightarrow M$', '$E \longrightarrow E + E$'] #['Nact', 'NM', 'EM', 'Ediv']
    reg_or = [Nact_reg[0:2]] + [NM_reg[0:2]] + [EM_reg[0:2]] + [Ediv_reg[0:2]]
    reg_and = [Nact_reg[2]] + [NM_reg[2]] + [EM_reg[2]] + [Ediv_reg[2]]
    reg_and_label = [param_names[-14]] + [param_names[-10]] + [param_names[-6]] + [param_names[-2]]
    reg_bl = [Nact_reg[3]] + [NM_reg[3]] + [EM_reg[3]] + [Ediv_reg[3]]
    reg_bl_label = [param_names[-13]] + [param_names[-9]] + [param_names[-5]] + [param_names[-1]]
else:
    reg = [r for r in ((Nact_reg if "Nact" in comment else []) + (NM_reg if "NM" in comment else []) + (EM_reg if "EM" in comment else []) + (Ediv_reg if "Ediv" in comment else [])) if np.std(infection_scenarios[0][r].values) > 0]
    reg_label = [r for i, r in enumerate(param_names[-16:])  if  (Nact_reg + NM_reg + EM_reg + Ediv_reg)[i] in reg] 
    modules = (['$N \longrightarrow N^*$'] if "Nact" in comment else []) + (['$N^* \longrightarrow M$'] if "NM" in comment else []) + (['$E \longrightarrow M$'] if "EM" in comment else []) + (['$E \longrightarrow E + E$'] if "Ediv" in comment else [])
    reg_or = (Nact_reg[0:2] if "Nact" in comment else []) + (NM_reg[0:2] if "NM" in comment else []) + (EM_reg[0:2] if "EM" in comment else []) + (Ediv_reg[0:2] if "Ediv" in comment else [])
    reg_and = ([Nact_reg[2]] if "Nact" in comment else []) + ([NM_reg[2]] if "NM" in comment else []) + ([EM_reg[2]] if "EM" in comment else []) + ([Ediv_reg[2]] if "Ediv" in comment else [])
    reg_and_label = ([param_names[-14]] if "Nact" in comment else []) + ([param_names[-10]] if "NM" in comment else []) + ([param_names[-6]] if "EM" in comment else []) + ([param_names[-2]] if "Ediv" in comment else [])
    reg_bl = ([Nact_reg[3]] if "Nact" in comment else []) + ([NM_reg[3]] if "NM" in comment else []) + ([EM_reg[3]] if "EM" in comment else []) + ([Ediv_reg[3]] if "Ediv" in comment else [])
    reg_bl_label = ([param_names[-13]] if "Nact" in comment else []) + ([param_names[-9]] if "NM" in comment else []) + ([param_names[-5]] if "EM" in comment else []) + ([param_names[-1]] if "Ediv" in comment else [])

num_categories = len(perf_labels)
reg_markers = ['o', 's', 'v', 'X', '^']
num_modules = len(modules)
num_psis = len(Nact_reg)

key_var = 'antigenicity_over_harm'
key_var_label = "Ag.-inflam. salience\n"+r"$\frac{\tau_I^{-1}}{\tau_H^{-1}}$"
```

```python
# Find the best performing architectures within network modules
top_perf_median_list = []
top_perf_extrema_list = []


for i, perf in enumerate(perf_vars):
    if 'timing' in perf or 'toxicity' in perf or 'clearance' in perf:
        top_perf_median = clustered_mean_df.groupby(vir_vars, as_index=False).median()
        
        id_extrema = clustered_mean_df.groupby(vir_vars)[perf].idxmin()
        top_perf_extrema = clustered_mean_df.loc[id_extrema] #.loc[clustered_mean_df[perf] == 1].groupby(vir_vars, as_index=False).min()
    else:
        top_perf_median = clustered_mean_df.groupby(vir_vars, as_index=False).median()
        
        id_extrema = clustered_mean_df.groupby(vir_vars)[perf].idxmax()
        top_perf_extrema = clustered_mean_df.loc[id_extrema]
        
    top_perf_median['perf_index'] = i
    top_perf_median_list.append(top_perf_median)
    top_perf_extrema['perf_index'] = i
    top_perf_extrema_list.append(top_perf_extrema)

top_perf_median_df = pd.concat(top_perf_median_list)
top_perf_median_df['antigenicity_over_harm'] = antigenicity_over_harm(top_perf_median_df)
top_perf_median_df.sort_values(by=key_var, inplace=True)
top_perf_extrema_df = pd.concat(top_perf_median_list)
top_perf_extrema_df['antigenicity_over_harm'] = antigenicity_over_harm(top_perf_extrema_df)
top_perf_extrema_df.sort_values(by=key_var, inplace=True)
```

### (b) Impact of cell fate modules on the immune response

```python
# Impactfulness of networks
clustered_mean_df['antigenicity_over_harm'] = antigenicity_over_harm(clustered_mean_df)

x_varname= key_var
fig_j, axs = plt.subplots(num_categories, num_modules, figsize=(num_modules*np.maximum(3.5, num_modules), num_categories*np.maximum(3.5, num_modules)), sharex='row', sharey='row')

for y in np.arange(0, num_modules):
    reg_choice_label = modules[y]
    reg_choice = module_reg[reg_choice_label]
    vir_data_mean = clustered_mean_df[reg + vir_vars + [x_varname] + perf_vars].groupby(vir_vars + [x_varname] + reg_choice, as_index=False).mean()
    data_std = vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).std()
    data_mean = np.abs(vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).mean())
    data_max = vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).max() - data_mean
    data_min = data_mean - vir_data_mean.groupby(vir_vars + [x_varname], as_index=False).max()
    
    
    for i, category in enumerate(perf_labels):
        y_varname = perf_vars[i]

        im = axs[i, y].scatter(data_std[x_varname], data_std[y_varname]/data_mean[y_varname], 
                              marker ='s', s = 50, c = data_mean['b_I']*data_mean['S_0'] - data_mean['d_I'], cmap='coolwarm_r', edgecolors = 'None', norm=mpl.colors.Normalize())
        axs[i, y].scatter(data_mean[x_varname], np.maximum(data_max[y_varname],data_min[y_varname])/data_mean[y_varname], marker = '_', 
                            s = 50, c = data_mean['b_I']*data_mean['S_0'] - data_mean['d_I'], cmap='coolwarm_r', linewidths=2.5, label = r'Largest observed effect' if (i == 0)*(y == num_modules -1) else '')
        #axs[i, y].semilogx()
        axs[i, y].semilogy()
        axs[i, y].yaxis.set_tick_params(labelbottom=True)
        
        axs[i, y].minorticks_off()
        axs[i, y].tick_params(axis='both', which='major', labelsize=8)
        axs[i, y].set_box_aspect(1)
        axs[i, y].set_title(perf_labels[i]+'\n'+ modules[y], fontsize = 10.0)
        cb = fig_j.colorbar(im, ax=axs[i, y], orientation='vertical', shrink = 0.75, pad = 0.05)
        cb.set_label("Reproduction rate, \n"+r"$b_I\times S(0) - d_I$", fontsize = 10)
        cb.ax.tick_params(labelsize=8)
        

axs[0,-1].legend(loc='center right', bbox_to_anchor=(3.5, 0.5), fontsize = 12)
fig_j.subplots_adjust(wspace=.65, hspace=.85)

for k, ax in enumerate(axs.flat):
    ax.set_xlabel(key_var_label, fontsize = 12)
    ax.set_ylabel(r"$\frac{\text{Standard deviation}}{\text{Mean}}$", fontsize = 12)
        
plt.savefig('_figs/'+comment+'-modularity_by_performance_w_std'+'.png', dpi=150, bbox_inches="tight")
```

### (d) Dependence of performance on modules and architectures

```python
# Define utility and constraint functions
from scipy.optimize import minimize, Bounds, NonlinearConstraint

def outcome(x, antigenicity_over_harm = 0.0, K_EH = K_EH, perf_index = 0, type = "average"):

    xs = x.reshape(-1,len(x))
    if type == "average":
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data['K_EH'] == K_EH)]
    else:
        input = impact_data.loc[(impact_data.perf_index == perf_index)*(impact_data.antigenicity_over_harm == antigenicity_over_harm)*(impact_data['K_EH'] == K_EH)]
        
    out = np.array([sigmoid(xs, *p).item() for p in input[['y_max', 'y_min']+[('intercept_' + j) for j in reg] + reg].values])
    return out

def utility(x, antigenicity_over_harm = 0.0, K_EH = K_EH, lam = 1, type = "average"):
    if type == "average":
        out = np.mean(outcome(x, 0.0, K_EH, 0, type = type) - lam*outcome(x, 0.0, K_EH, 1, type = type))
    else:
        out = outcome(x, antigenicity_over_harm, K_EH, 0, type = type) - lam*outcome(x, antigenicity_over_harm, K_EH, 1, type = type)

    return -out

def constraint1(x, bound = psi_max):
    return np.sum(np.abs(x[0:4])) - bound

def constraint2(x, bound = psi_max):
    return np.sum(np.abs(x[4:8])) - bound

def constraint3(x, bound = psi_max):
    return np.sum(np.abs(x[8:12])) - bound

def constraint4(x, bound = psi_max):
    return np.sum(np.abs(x[12:16])) - bound

def constraint5(x, bound = psi_max):
    return np.max(np.abs(x)) - bound

# define constraints
const1 = NonlinearConstraint(constraint1, -np.inf, 0)
const2 = NonlinearConstraint(constraint2, -np.inf, 0)
const3 = NonlinearConstraint(constraint3, -np.inf, 0)
const4 = NonlinearConstraint(constraint4, -np.inf, 0)
const5 = NonlinearConstraint(constraint5, -np.inf, 0)
```

```python
# Optimize average utlity
import warnings
warnings.filterwarnings('ignore') # suppress warnings 

lams = np.linspace(0, 5, 11)
avg_util_psi_opts = []
avg_util_optimum = []

for l in tqdm(lams):
    res = minimize(utility, args = (0.0, K_EH, l),
                   x0 = np.zeros(len(reg)), 
                   tol=1e-6,
                   constraints = [const1, const2, const3, const4])
    avg_util_psi_opts.append(res.x)
    avg_util_optimum.append([-utility(res.x, 0.0, K_EH, l)])

# reshape data
avg_util_optimum_data = np.hstack(avg_util_optimum)
avg_util_psi_opts_data = np.vstack(avg_util_psi_opts)
plt.scatter(lams, avg_util_optimum_data)
```

```python
# Regress Performance measures on regulation parameters
num_cpu = 150

def sigmoid(x, y_max, y_min, b0, b1, b2, b3, b4, b5, b6, b7, w0, w1, w2, w3, w4, w5, w6, w7, a00, a01):

    out = y_max*((1 - a00)/(1 + np.exp( -((b0 + x[:,0])*w0 + (b1 + x[:,1])*w1 + (b2 + x[:,2])*w2 + (b3 + x[:,3])*w3) )) + a00)*((1 - a01)/(1 + np.exp( -((b4 + x[:,4])*w4 + (b5 + x[:,5])*w5 + (b6 + x[:,6])*w6 + (b7 + x[:,7])*w7) )) + a01) + y_min

    return out  

def model(input, fit = "linear"):

    out = []
    infection = input[vir_vars].values[0]
    for i, perf in enumerate(auto_perf_vars):
        # run regression
        
        if fit == "linear":
            input.loc[:,'dep_var'] = np.log(1 + input[perf].values) if 'Memory' in auto_perf_labels[auto_perf_vars.index(perf)] or 'magnitude' in auto_perf_labels[auto_perf_vars.index(perf)] else input[perf].values
            lr = smf.ols('dep_var ~ ' + ' + '.join(reg), data=input).fit()
    
            # save values
            out.append( (lr.params.values[0: 1 + len(reg)]).tolist() + np.sqrt(np.diag(lr.cov_params())).tolist() + infection.tolist() + [np.mean(input[perf].values), i] )
        else:
            y_max = np.max(input[perf].values)
            y_min = np.min(input[perf].values)
            y_std = np.std(input[perf].values)
            popt, pcov = curve_fit(sigmoid, input[reg].values, input[perf].values, 
                           p0 =[y_max, 0.0 if y_min > 0.0 else y_min] + len(reg)*[0.0] + (len(reg))*[0.0] + [0.0, 0.0],
                           method = 'trf', 
                           bounds = ([y_max - y_std, 0.0 if y_min >= 0.0 else y_min - y_std] + len(reg)*[-2.5*psi_max] + (len(reg))*[-2.5*psi_max] + [0.0, 0.0], 
                                     [10*y_max if y_max > 0 else y_std + 0.1, y_max if y_min > 0 else y_std  + 0.1] + len(reg)*[2.5*psi_max] + (len(reg))*[2.5*psi_max] + [1.0, 1.0]), 
                                   maxfev=10000)
            
            residuals = input[perf].values - sigmoid(input[reg].values, *popt)
            rmse = np.sqrt(np.mean(residuals**2))
            out.append(popt.tolist() + np.sqrt(np.diag(pcov)).tolist() + infection.tolist() + [np.mean(input[perf].values), np.max(input[perf].values), np.min(input[perf].values), np.median(input[perf].values), i, rmse, y_std])
    
    return np.vstack(out)

# # Compile datasets
auto_impact_data = pd.DataFrame(np.vstack(Parallel(n_jobs = num_cpu, batch_size = max(int(len(autos)/num_cpu),1))(delayed(model)(input = data.groupby(vir_vars + reg, as_index=False).mean(), fit = "nonlinear") 
                                                                                                            for data in auto_scenarios)), 
                               columns = ['y_max', 'y_min'] + [('intercept_' + j) for j in reg] + reg + ['unreg_frac1', 'unreg_frac1'] + ['sigma_y_max', 'sigma_y_min'] + [('sigma_intercept_' + j) for j in reg] + ['sigma_'+r  for r in (reg + ['unreg_frac1', 'unreg_frac1'])] + vir_vars + ['mean_perf', 'max_perf', 'min_perf', 'median_perf', 'perf_index', 'rmse', 'y_std'])

auto_impact_data.to_pickle('/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/summary_stats/mean/'+"auto-Nact-Ediv-vir"+'-'+'impact_data'+'.pkl')

# Import dataset
auto_impact_data = pd.read_pickle('/gscratch/scrubbed/oukogu/infoimmune/sim_output/no_cell_var/summary_stats/mean/'+"auto-Nact-Ediv-vir"+'-'+'impact_data'+'.pkl')  
```

```python
# Regulation of networks versus performance
fig_h, axs = plt.subplots(1, 2, figsize=(4, 3.0), sharex=True, sharey=True)

# compute risk-utility
utility_df = clustered_mean_df.groupby(reg, as_index=False).mean() #/np.mean(clustered_mean_df.loc[(np.max(np.abs(clustered_mean_df[reg]), axis = 1) == 0)][var])
utility_df.sort_values(by=reg, inplace=True)
risk_df = clustered_mean_df.groupby(reg, as_index=False).std()
risk_df.sort_values(by=reg, inplace=True)

auto_virs = np.unique(autoreactive_mean_df[['d_S','d_I','K_IE', 'b_I']].to_numpy(), axis = 0)
auto_risk_data = autoreactive_mean_df.groupby(['d_S','d_I','K_IE', 'b_I'] + reg, as_index=False).mean()
auto_risk_data.sort_values(by=reg, inplace=True)

for p, cond in enumerate(auto_virs):
    indices = [2,3,6,7,10,11,14,15]
    x = utility_df[auto_perf_vars[0]]
    y = risk_df[auto_perf_vars[0]]/np.abs(utility_df[auto_perf_vars[0]])

    z = auto_risk_data.loc[(np.sum(auto_risk_data[['d_S','d_I','K_IE', 'b_I']] - auto_virs[p], axis = 1) == 0)][auto_perf_vars[1]]
    im = axs[p].scatter(x, y, marker ='o', s = 2.5, c = z, cmap='coolwarm', edgecolors = 'None', alpha = 0.5,
                              norm = mpl.colors.Normalize(vmin= np.min(z), vmax= np.max(z)))
    axs[p].set_yscale("log")
    axs[p].set_xscale("log")
    axs[p].set_box_aspect(1)

    # make inset figure with top 95%
    cutoff = np.quantile(utility_df[auto_perf_vars[0]], 0.95)
    xins = utility_df.loc[utility_df[auto_perf_vars[0]] >= cutoff][auto_perf_vars[0]]
    yins = risk_df.loc[utility_df[auto_perf_vars[0]] >= cutoff][auto_perf_vars[0]]/np.abs(xins)
    zins = utility_df.loc[utility_df[auto_perf_vars[0]] >= cutoff][var]

    axins = inset_axes(axs[p], width="30%", height="40%", loc=3)
    axins.scatter(xins, yins, marker ='o', s = 2.5, c = zins, cmap='coolwarm', edgecolors = 'None', alpha = 0.35,
                              norm = mpl.colors.Normalize(vmin= np.min(z), vmax= np.max(z)))
    axins.tick_params(labelleft=False, labelbottom=False)
    
    cb = fig_h.colorbar(im, ax=axs[p], orientation='horizontal', shrink = 1.0, pad = 0.3)
    cb.set_label(perf_labels[1]+"\n"+r"$d_S =$ "+str(auto_virs[p,0] + auto_virs[p,1])+",\n"+r"$K_{S,E} =$ "+str(int(auto_virs[p,2])), fontsize = 10)
                
fig_h.subplots_adjust(wspace=.3, hspace=0.65)

for k, ax in enumerate(axs.flat):
    if k == 0:
        ax.set_ylabel("Coef.\n of variation", fontsize = 12)
        ax.set_xlabel("Average utility", fontsize = 12)
        #ax.semilogx()
        
plt.savefig('_figs/'+"auto-Nact-Ediv-vir"+'-'+'_utility_v_risk_by_auto_sctr.png', dpi=150, bbox_inches="tight")
del utility_df
del risk_df
```
